## Part A: Python syntax for business analysis

Python works with values, variables and expressions. A variable is simply a named container for a value.


```python
revenue = 250000
cost = 205000
profit = revenue - cost
profit_margin_pct = profit / revenue * 100

print("Revenue:", revenue)
print("Profit:", profit)
print("Profit margin (%):", round(profit_margin_pct, 2))
```

### Common business data types

- `int`: whole numbers, such as units sold
- `float`: decimal values, such as discount rates
- `str`: text, such as region names
- `bool`: `True` or `False`, such as whether a target was achieved


```python
units_sold = 125                 # integer
average_price = 1499.50          # float
region = "West"                  # string
target_achieved = True           # Boolean

print(type(units_sold))
print(type(average_price))
print(type(region))
print(type(target_achieved))
```

### Lists: storing a sequence of values

A list is useful for a small collection, such as monthly sales. Python uses zero-based indexing, so the first item is at position `0`.


```python
monthly_sales = [120000, 135000, 128000, 150000]

print("First month:", monthly_sales[0])
print("Number of months:", len(monthly_sales))
print("Total sales:", sum(monthly_sales))
print("Highest monthly sales:", max(monthly_sales))
print("Average monthly sales:", sum(monthly_sales) / len(monthly_sales))
```

### Dictionaries: storing labelled business values

A dictionary connects a key with a value, which is useful for simple lookup-style information.


```python
regional_sales = {
    "North": 520000,
    "South": 610000,
    "East": 430000,
    "West": 690000
}

print("West sales:", regional_sales["West"])
print("Regions:", list(regional_sales.keys()))
print("Sales values:", list(regional_sales.values()))
```

### Conditional logic

Conditions allow business rules to be expressed in code.


```python
actual_profit = 185000
target_profit = 175000

if actual_profit >= target_profit:
    print("Profit target achieved")
else:
    print("Profit target not achieved")
```

### Repeating an operation with a loop

Loops are useful for small repeated tasks. For transactional data, Pandas will soon provide a more efficient approach.


```python
regional_sales = {
    "North": 520000,
    "South": 610000,
    "East": 430000,
    "West": 690000
}

for region_name, sales_value in regional_sales.items():
    if sales_value >= 600000:
        status = "High-performing"
    else:
        status = "Requires attention"
    print(region_name, ":", status)
```

## Part C: Selecting and filtering business data

The square-bracket syntax selects columns. A single column returns a Series; multiple columns require a list of column names.


```python
# One column
revenue_series = df["Revenue_INR"]
print(revenue_series.head())

```

```python
# Multiple columns
display(df[["Transaction_ID", "Region", "Revenue_INR", "Profit_INR"]].head())
```

### Select rows by position with `iloc`

`iloc` is useful when you want rows or columns by numerical position.


```python
# First five rows and first four columns
df.iloc[0:5, 0:4]
```

### Filter using one condition

The expression inside the brackets creates a `True`/`False` mask for every row.


```python
west_transactions = df[df["Region"] == "West"]
print("West transactions:", west_transactions.shape[0])
west_transactions.head()
```

### Filter using multiple conditions

Use `&` for AND, `|` for OR, and place each condition inside parentheses.


```python
west_high_profit = df[
    (df["Region"] == "West") &
    (df["Profit_INR"] > 3000)
]

west_high_profit[["Transaction_ID", "Store", "Product_Category", "Revenue_INR", "Profit_INR"]].head(10)
```

### Filter categories using `isin()`

`isin()` avoids writing several OR conditions.


```python
selected_categories = ["Electronics", "Furniture"]
high_value_categories = df[df["Product_Category"].isin(selected_categories)] ## helps to filter rows.
high_value_categories[["Product_Category", "Revenue_INR", "Profit_INR"]].head()
```

```python
print(high_value_categories.shape)
```

### Use `query()` for readable business conditions

`query()` can make multi-condition filters easier to read.


```python
df.query("Region == 'South' and Revenue_INR > 20000")[[
    "Store", "Product_Category", "Revenue_INR", "Profit_INR"
]].head()

# .isin() checks whether values belong to a list of specified values.
# .query() filters rows using a condition written as a string expression.
```

### Create derived business metrics

Derived columns convert raw variables into useful measures. Here, we create a calculated margin and classify transactions by revenue.


```python
df["Calculated_Margin_Pct"] = df["Profit_INR"] / df["Revenue_INR"] * 100

df.head()
```

```python
df["Revenue_Band"] = pd.cut(
    df["Revenue_INR"],
    bins=[0, 5000, 15000, 30000, np.inf],
    labels=["Low", "Medium", "High", "Very High"]
)

df.head()
```

### Extract date components

Date components support monthly, quarterly or day-level analysis.


```python
df["Transaction_Date"].head()
```

```python
df["Year"] = df["Transaction_Date"].dt.year
df["Month"] = df["Transaction_Date"].dt.month
df["Month_Name"] = df["Transaction_Date"].dt.month_name().str[:3]
df["Quarter"] = df["Transaction_Date"].dt.to_period("Q").astype(str)

df[["Transaction_Date", "Year", "Month", "Month_Name", "Quarter"]].head()
```

## Part D: Answering business queries with Pandas

Sorting and ranking help identify top or bottom records. Always confirm what the ranking metric represents.


```python
top_revenue_transactions = df.nlargest(10, "Revenue_INR")
top_revenue_transactions[[
    "Transaction_ID", "Store", "Region", "Product_Category", "Revenue_INR", "Profit_INR"
]]
```

### Query 1: What are the basic business totals?

Pandas aggregations such as `sum`, `mean`, `median`, `min` and `max` provide quick answers.


```python
print("Total revenue:", df["Revenue_INR"].sum())
print("Total profit:", df["Profit_INR"].sum())
print("Average revenue per row:", df["Revenue_INR"].mean())
print("Median revenue per row:", df["Revenue_INR"].median())
print("Largest transaction:", df["Revenue_INR"].max())
```

### Query 2: Which region generates the most revenue?

`groupby()` divides rows into groups, and an aggregation summarises each group.


```python
revenue_by_region = (
    df.groupby("Region", as_index=False)["Revenue_INR"]
      .sum()
      .sort_values("Revenue_INR", ascending=False)
)
revenue_by_region
```

### Query 3: Which product categories generate revenue and profit?

Use `agg()` when more than one metric or aggregation is required.


```python
category_performance = (
    df.groupby("Product_Category", as_index=False)
      .agg(
          Transactions=("Transaction_ID", "count"),
          Units=("Quantity", "sum"),
          Revenue_INR=("Revenue_INR", "sum"),
          Profit_INR=("Profit_INR", "sum"),
          Average_Discount_Pct=("Discount_Pct", "mean")
      )
)
category_performance["Profit_Margin_Pct"] = (
    category_performance["Profit_INR"] / category_performance["Revenue_INR"] * 100
)
category_performance.sort_values("Profit_INR", ascending=False)
```

### Query 4: Which category performs best in each region?

Grouping by two variables creates a more detailed business view.


```python
region_category = (
    df.groupby(["Region", "Product_Category"], as_index=False)
      .agg(Revenue_INR=("Revenue_INR", "sum"), Profit_INR=("Profit_INR", "sum"))
)
region_category["Profit_Margin_Pct"] = region_category["Profit_INR"] / region_category["Revenue_INR"] * 100
region_category.sort_values(["Region", "Profit_INR"], ascending=[True, False]).head(15)
```

### Pivot tables for managerial summaries

A pivot table places categories on rows and regions on columns, making comparisons easier.


```python
profit_pivot = pd.pivot_table(
    df,
    index="Product_Category",
    columns="Region",
    values="Profit_INR",
    aggfunc="sum",
    fill_value=0,
    margins=True
)
profit_pivot
```

## Part E: Data quality before EDA

Analytical outputs are only as reliable as the underlying data. We will inspect missing values, duplicates and unusual values before drawing conclusions.


```python
data_quality = pd.DataFrame({
    "Data_Type": df.dtypes.astype(str),
    "Missing_Count": df.isna().sum(),
    "Missing_Pct": (df.isna().mean() * 100).round(2),
    "Unique_Values": df.nunique(dropna=True)
})

data_quality
```

### Inspect duplicates and create an analysis copy

Do not alter the original DataFrame immediately. Create a cleaned copy so the treatment remains transparent.


```python
print("Exact duplicate rows:", df.duplicated().sum())

duplicate_examples = df[df.duplicated(keep=False)].sort_values("Transaction_ID")
display(duplicate_examples.head(10))

analysis_df = df.drop_duplicates().copy()
print("Original shape:", df.shape)
print("After removing exact duplicates:", analysis_df.shape)
```

### Missing-value treatment

Treatment depends on meaning. Here, we retain unknown customer segments as an explicit category and leave numerical missing values for analyses that can safely ignore them.


```python
analysis_df["Customer_Segment"] = analysis_df["Customer_Segment"].fillna("Unknown")

print("Remaining missing values:")
print(analysis_df.isna().sum()[analysis_df.isna().sum() > 0])
```
